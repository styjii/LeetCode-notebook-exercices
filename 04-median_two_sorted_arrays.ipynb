```python
import random
import time
from IPython.display import HTML, display

# Les cas courts ont des resultats attendus ecrits explicitement.
cas_tests = [
    ("Exemple 1", [1, 3], [2], 2.0),
    ("Exemple 2", [1, 2], [3, 4], 2.5),
    ("nums1 vide", [], [1], 1.0),
    ("nums2 vide", [2], [], 2.0),
    ("nums1 vide, total pair", [], [2, 3], 2.5),
    ("Zeros", [0, 0], [0, 0], 0.0),
    ("Un element chacun", [1], [2], 1.5),
    ("Tableaux disjoints (nums1 < nums2)", [1, 2, 3], [4, 5, 6], 3.5),
    ("Tableaux disjoints (nums2 < nums1)", [7, 8, 9, 10], [1, 2], 7.5),
    ("Tailles tres differentes", [1, 2, 3, 4, 5], [6], 3.5),
    ("Doublons", [1, 1, 1], [1, 1, 2], 1.0),
    ("Valeurs negatives", [-5, -3, -1], [-4, -2], -3.0),
    ("Valeurs extremes", [-10**6, 0], [0, 10**6], 0.0),
    ("Entrelaces", [1, 3, 5, 7], [2, 4, 6, 8], 4.5),
]


def mediane_reference(a, b):
    """Reference de verification (tri complet) : sert uniquement a controler vos resultats."""
    s = sorted(a + b)
    k = len(s)
    return float(s[k // 2]) if k % 2 else (s[k // 2 - 1] + s[k // 2]) / 2


# Cas aleatoires, y compris aux tailles maximales des contraintes
random.seed(2026)
for k, (m, n) in enumerate([(1000, 1000), (1000, 1), (1, 1000), (999, 1000), (500, 1500 - 500)], 1):
    a = sorted(random.randint(-10**6, 10**6) for _ in range(m))
    b = sorted(random.randint(-10**6, 10**6) for _ in range(n))
    cas_tests.append((f"Aleatoire m={m}, n={n}", a, b, mediane_reference(a, b)))

sol = Solution()
cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes, reussis = [], 0


def apercu(v):
    return str(v) if len(v) <= 8 else str(v[:4])[:-1] + f", ... ({len(v)} elem.)]"


def ligne(nom, entree, attendu, obtenu, ok):
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    return (f"<tr><td style='{cell}'>{nom}</td>"
            f"<td style='{cell}font-family:monospace;'>{entree}</td>"
            f"<td style='{cell}font-family:monospace;'>{attendu}</td>"
            f"<td style='{cell}font-family:monospace;'>{obtenu}</td>"
            f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;"
            f"border-radius:12px;font-size:12px;font-weight:600;'>{etat}</span></td></tr>")


for nom, a, b, attendu in cas_tests:
    try:
        obtenu = sol.findMedianSortedArrays(list(a), list(b))
        ok = isinstance(obtenu, (int, float)) and abs(obtenu - attendu) < 1e-5
    except Exception as erreur:
        obtenu, ok = f"Erreur : {erreur}", False
    reussis += ok
    lignes.append(ligne(nom, f"nums1={apercu(a)}<br>nums2={apercu(b)}", attendu, obtenu, ok))

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"

# ---- Test bonus (hors contraintes) : 2 tableaux de 1 000 000 elements ----
N = 1_000_000
grand1 = list(range(0, 2 * N, 2))      # 0, 2, 4, ...
grand2 = list(range(1, 2 * N, 2))      # 1, 3, 5, ...
attendu_bonus = N - 0.5                # fusion = 0, 1, ..., 2N-1
debut = time.perf_counter()
try:
    obtenu_bonus = sol.findMedianSortedArrays(grand1, grand2)
    bonus_ok = isinstance(obtenu_bonus, (int, float)) and abs(obtenu_bonus - attendu_bonus) < 1e-5
except Exception as erreur:
    obtenu_bonus, bonus_ok = f"Erreur : {erreur}", False
duree_bonus = time.perf_counter() - debut

if not bonus_ok:
    msg_bonus, couleur_b = "Resultat incorrect sur les tres grands tableaux.", "#c0392b"
elif duree_bonus < 0.01:
    msg_bonus, couleur_b = "Temps compatible avec une complexite logarithmique O(log(m + n)).", "#2e9e5b"
else:
    msg_bonus, couleur_b = ("Resultat correct, mais temps compatible avec une complexite lineaire ou pire. "
                            "Objectif : O(log(m + n)).", "#d9892b")

th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Entree</th><th style="{th}">Attendu</th>
<th style="{th}">Obtenu</th><th style="{th}">Statut</th></tr>
{''.join(lignes)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div>
<div style="margin-top:10px;padding:10px 16px;border-left:4px solid {couleur_b};background:rgba(128,128,128,0.08);line-height:1.6;">
<b>Test bonus (2 x 1 000 000 elements) :</b> {duree_bonus*1000:.2f} ms<br>{msg_bonus}
</div></div>"""))
```
