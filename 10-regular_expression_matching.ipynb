<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">1. &Eacute;nonc&eacute;</span>
</div>
<div style="border-left:4px solid #2f6fad;background:rgba(47,111,173,0.08);padding:14px 20px;border-radius:0 8px 8px 0;line-height:1.7;">
&Eacute;tant donn&eacute; une cha&icirc;ne d'entr&eacute;e <code>s</code> et un motif <code>p</code>, impl&eacute;mentez la correspondance d'expression r&eacute;guli&egrave;re avec la prise en charge de <code>'.'</code> et <code>'*'</code> :
<ul style="margin:8px 0 0 0;">
<li><code>'.'</code> correspond &agrave; <b>n'importe quel caract&egrave;re unique</b>.</li>
<li><code>'*'</code> correspond &agrave; <b>z&eacute;ro ou plusieurs occurrences</b> de l'&eacute;l&eacute;ment pr&eacute;c&eacute;dent.</li>
</ul>
</div>
<div style="margin-top:14px;line-height:1.7;">
Retournez un bool&eacute;en indiquant si la correspondance couvre <b>toute</b> la cha&icirc;ne <code>s</code> (et non une partie seulement).
</div>
<div style="margin-top:14px;display:flex;flex-wrap:wrap;gap:12px;"><div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Correspondance totale</div>
<div style="margin-top:4px;line-height:1.6;">Le motif doit consommer la cha&icirc;ne enti&egrave;re : &laquo; a &raquo; ne correspond pas &agrave; &laquo; aa &raquo;.</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Op&eacute;rateur *</div>
<div style="margin-top:4px;line-height:1.6;">Il s'applique &agrave; l'&eacute;l&eacute;ment qui le <b>pr&eacute;c&egrave;de</b> (une lettre ou un point), jamais seul.</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Sortie</div>
<div style="margin-top:4px;line-height:1.6;">Un bool&eacute;en (<code>True</code> ou <code>False</code> en Python).</div></div>
</div>
