```python
import sqlite3
from IPython.display import HTML, display


def creer_base(salaires):
    """Cree une base SQLite en memoire avec la table Employee(id, salary).
    `salaires` est la liste des salaires ; les identifiants sont 1, 2, 3, ..."""
    conn = sqlite3.connect(":memory:")
    conn.execute("CREATE TABLE Employee (id INTEGER PRIMARY KEY, salary INTEGER)")
    conn.executemany("INSERT INTO Employee VALUES (?, ?)", list(enumerate(salaires, 1)))
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange)."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{v}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1 (pour l'exemple 2, utilisez creer_base([100]))
base_exemple = creer_base([100, 200, 300])

afficher(*executer("SELECT * FROM Employee", base_exemple), titre="Table Employee")
```

```python
import html

COLONNE_ATTENDUE = "SecondHighestSalary"

# Chaque cas : (nom, liste des salaires, deuxieme salaire distinct attendu ou None)
cas_tests = [
    ('Exemple 1', [100, 200, 300], 200),
    ('Exemple 2 : un seul salaire', [100], None),
    ('Table vide', [], None),
    ('Doublon sur le maximum', [300, 300, 200], 200),
    ('Tous les salaires identiques', [100, 100, 100], None),
    ('Deux salaires distincts', [100, 200], 100),
    ('Deux lignes au meme salaire', [200, 200], None),
    ('Salaires non tries avec doublons', [500, 100, 300, 300, 500], 300),
    ('Le deuxieme salaire vaut zero', [0, 0, 50], 0),
    ('Salaires negatifs', [-5, -10, -5, -1], -5),
    ('Valeurs tres grandes', [2147483647, 2147483646, 2147483647], 2147483646),
    ('Doublon sur le deuxieme salaire', [900, 700, 700, 700, 100], 700)
]

# Grand jeu de donnees : 1000 lignes, salaires de 0 a 9 repetes (le deuxieme plus haut est 8)
cas_tests.append(("Grand jeu : 1000 lignes avec nombreux doublons", [i % 10 for i in range(1000)], 8))

cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def format_val(v):
    return "NULL" if v is None else str(v)


for nom, salaires, attendu in cas_tests:
    conn = creer_base(salaires)
    detail, obtenu_txt = "", "-"
    try:
        colonnes, lignes = executer(requete, conn)
        if colonnes != [COLONNE_ATTENDUE]:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendue : ['{COLONNE_ATTENDUE}']"
        elif len(lignes) != 1:
            ok = False
            obtenu_txt = f"{len(lignes)} ligne(s)"
            detail = ("Aucune ligne retournee : il faut retourner une ligne contenant NULL."
                      if len(lignes) == 0 else "Le resultat doit contenir exactement une ligne.")
        else:
            valeur = lignes[0][0]
            obtenu_txt = format_val(valeur)
            ok = valeur == attendu and (valeur is None) == (attendu is None)
            if not ok:
                detail = f"Valeur obtenue : {html.escape(format_val(valeur))} ; attendue : {format_val(attendu)}"
    except Exception as erreur:
        ok, obtenu_txt, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    apercu = str(salaires) if len(salaires) <= 8 else f"{len(salaires)} salaires"
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}font-family:monospace;'>{apercu}</td>"
        f"<td style='{cell}font-family:monospace;'>{format_val(attendu)}</td>"
        f"<td style='{cell}font-family:monospace;'>{obtenu_txt}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Salaires</th><th style="{th}">Attendu</th>
<th style="{th}">Obtenu</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
