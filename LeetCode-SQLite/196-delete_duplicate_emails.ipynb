```python
import sqlite3
from IPython.display import HTML, display


def creer_base(lignes):
    """Cree une base SQLite en memoire avec la table Person(id, email).
    `lignes` est une liste de tuples (id, email)."""
    conn = sqlite3.connect(":memory:")
    conn.execute("CREATE TABLE Person (id INTEGER PRIMARY KEY, email TEXT NOT NULL)")
    conn.executemany("INSERT INTO Person VALUES (?, ?)", lignes)
    conn.commit()
    return conn


def executer(requete, conn):
    """Execute l'instruction (une seule) et valide la transaction. Retourne le nombre de lignes modifiees."""
    curseur = conn.execute(requete)
    conn.commit()
    return curseur.rowcount


def contenu(conn):
    """Retourne le contenu actuel de la table Person sous forme de liste de tuples (id, email)."""
    return conn.execute("SELECT id, email FROM Person ORDER BY id").fetchall()


def afficher_table(conn, titre=""):
    """Affiche le contenu de la table Person."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    lignes = contenu(conn)
    corps = "".join("<tr>" + "".join(f"<td style='{cellule}'>{v}</td>" for v in l) + "</tr>" for l in lignes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'><th style='{cellule}text-align:left;'>id</th>"
                 f"<th style='{cellule}text-align:left;'>email</th></tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
EXEMPLE = [(1, "john@example.com"), (2, "bob@example.com"), (3, "john@example.com")]
afficher_table(creer_base(EXEMPLE), titre="Table Person (exemple 1)")
```

```python
import html
import re
from collections import Counter

# Chaque cas : (nom, lignes de depart [(id, email)], contenu attendu de la table apres l'execution)
cas_tests = [
    ('Exemple 1',
     [(1, 'john@example.com'), (2, 'bob@example.com'), (3, 'john@example.com')],
     [(1, 'john@example.com'), (2, 'bob@example.com')]),
    ('Aucun doublon',
     [(1, 'a@b.com'), (2, 'c@d.com'), (3, 'e@f.com')],
     [(1, 'a@b.com'), (2, 'c@d.com'), (3, 'e@f.com')]),
    ('Adresse presente trois fois',
     [(1, 'x@y.com'), (2, 'x@y.com'), (3, 'z@y.com'), (4, 'x@y.com')],
     [(1, 'x@y.com'), (3, 'z@y.com')]),
    ('Plusieurs adresses en double',
     [(1, 'a@b.com'), (2, 'b@b.com'), (3, 'a@b.com'), (4, 'c@b.com'), (5, 'b@b.com'), (6, 'a@b.com')],
     [(1, 'a@b.com'), (2, 'b@b.com'), (4, 'c@b.com')]),
    ('Toutes les adresses identiques',
     [(1, 'a@b.com'), (2, 'a@b.com'), (3, 'a@b.com')],
     [(1, 'a@b.com')]),
    ('Une seule ligne',
     [(1, 'solo@mail.com')],
     [(1, 'solo@mail.com')]),
    ('Table vide',
     [],
     []),
    ('Doublons consecutifs',
     [(1, 'd@d.com'), (2, 'd@d.com'), (3, 'u@u.com')],
     [(1, 'd@d.com'), (3, 'u@u.com')]),
    ('Plus petit id insere en dernier',
     [(5, 'x@x.com'), (9, 'y@y.com'), (2, 'x@x.com'), (7, 'x@x.com')],
     [(9, 'y@y.com'), (2, 'x@x.com')]),
    ('Identifiants non consecutifs',
     [(10, 'a@a.com'), (200, 'b@b.com'), (3000, 'a@a.com'), (40, 'b@b.com')],
     [(10, 'a@a.com'), (40, 'b@b.com')]),
    ('Adresses proches mais differentes',
     [(1, 'a@b.com'), (2, 'a@b.co'), (3, 'a@bb.com'), (4, 'a@b.com')],
     [(1, 'a@b.com'), (2, 'a@b.co'), (3, 'a@bb.com')]),
    ('Adresses avec chiffres et symboles',
     [(1, 'a.b+1@mail.com'), (2, 'a.b+1@mail.com'), (3, 'a.b+2@mail.com'), (4, 'a_b@mail.com')],
     [(1, 'a.b+1@mail.com'), (3, 'a.b+2@mail.com'), (4, 'a_b@mail.com')])
]

# Grands jeux de donnees (1000 lignes)
cas_tests += [
    ("Grand jeu : 100 adresses, chacune presente 10 fois",
     [(i, f"u{i % 100}@x.com") for i in range(1, 1001)],
     [(i, f"u{i % 100}@x.com") for i in range(1, 101)]),
    ("Grand jeu : 1000 fois la meme adresse",
     [(i, "same@x.com") for i in range(1, 1001)], [(1, "same@x.com")]),
    ("Grand jeu : 1000 adresses toutes differentes",
     [(i, f"u{i}@x.com") for i in range(1, 1001)], [(i, f"u{i}@x.com") for i in range(1, 1001)]),
]


def est_delete(instruction):
    """Vrai si l'instruction (hors commentaires) commence par DELETE ou par une clause WITH."""
    sans_commentaires = re.sub(r"--[^\n]*", "", instruction)
    sans_commentaires = re.sub(r"/\*.*?\*/", "", sans_commentaires, flags=re.S)
    return sans_commentaires.strip().upper().startswith(("DELETE", "WITH"))


def court(liste, n=3):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0
instruction_valide = est_delete(requete)

for nom, depart, attendu in cas_tests:
    detail, nb_obtenu = "", "-"
    if not instruction_valide:
        ok = False
        detail = "La solution doit etre une instruction DELETE (et non un SELECT) ; aucune instruction DELETE trouvee."
    else:
        conn = creer_base(depart)
        try:
            executer(requete, conn)
            obtenu_lignes = [tuple(l) for l in contenu(conn)]
            nb_obtenu = len(obtenu_lignes)
            obtenu, voulu = Counter(obtenu_lignes), Counter(attendu)
            supprimees_a_tort = sorted((voulu - obtenu).elements(), key=str)
            non_supprimees = sorted((obtenu - voulu).elements(), key=str)
            ok = not supprimees_a_tort and not non_supprimees
            morceaux = []
            if supprimees_a_tort:
                morceaux.append(f"Lignes supprimees a tort (ou modifiees) : {court(supprimees_a_tort)}")
            if non_supprimees:
                morceaux.append(f"Lignes restantes en trop : {court(non_supprimees)}")
            detail = " | ".join(morceaux)
        except Exception as erreur:
            ok, detail = False, f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}'>{len(depart)}</td><td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Lignes avant</th><th style="{th}">Lignes attendues apres</th>
<th style="{th}">Lignes obtenues apres</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
