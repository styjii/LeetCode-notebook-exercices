```python
import sqlite3
from IPython.display import HTML, display


def creer_base(lignes):
    """Cree une base SQLite en memoire avec la table Activity(player_id, device_id, event_date, games_played).
    `lignes` est une liste de tuples (player_id, device_id, 'AAAA-MM-JJ', games_played)."""
    conn = sqlite3.connect(":memory:")
    # Le couple (player_id, event_date) est unique dans les donnees ; la contrainte n'est volontairement pas declaree
    # afin que SQLite ne cree pas d'index qui imposerait un ordre de lecture favorable aux requetes incorrectes.
    conn.execute("CREATE TABLE Activity (player_id INTEGER, device_id INTEGER, event_date TEXT NOT NULL, "
                 "games_played INTEGER)")
    conn.executemany("INSERT INTO Activity VALUES (?, ?, ?, ?)", lignes)
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange)."""
    cellule = "padding:6px 12px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{v}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base([
    (1, 2, "2016-03-01", 5),
    (1, 2, "2016-05-02", 6),
    (2, 3, "2017-06-25", 1),
    (3, 1, "2016-03-02", 0),
    (3, 4, "2018-07-03", 5),
])

afficher(*executer("SELECT * FROM Activity", base_exemple), titre="Table Activity")
```

```python
import html
from collections import Counter
from datetime import date, timedelta

COLONNES_ATTENDUES = ["player_id", "first_login"]

# Chaque cas : (nom, lignes de Activity [(player_id, device_id, event_date, games_played)],
#               lignes attendues [(player_id, first_login), ...])
cas_tests = [
    ('Exemple 1',
     [(1, 2, '2016-03-01', 5), (1, 2, '2016-05-02', 6), (2, 3, '2017-06-25', 1), (3, 1, '2016-03-02', 0), (3, 4, '2018-07-03', 5)],
     [(1, '2016-03-01'), (2, '2017-06-25'), (3, '2016-03-02')]),
    ('Un seul joueur, une seule connexion',
     [(1, 1, '2020-01-15', 3)],
     [(1, '2020-01-15')]),
    ('Dates non triees dans la table',
     [(1, 1, '2020-05-01', 2), (1, 1, '2020-01-01', 4), (1, 1, '2020-03-01', 1)],
     [(1, '2020-01-01')]),
    ('Plusieurs appareils pour un meme joueur',
     [(1, 10, '2020-02-01', 1), (1, 20, '2020-01-01', 2), (1, 30, '2020-03-01', 3), (1, 10, '2020-04-01', 4)],
     [(1, '2020-01-01')]),
    ('Premiere connexion sans partie jouee',
     [(1, 1, '2020-01-01', 0), (1, 1, '2020-01-02', 7)],
     [(1, '2020-01-01')]),
    ('Connexions a cheval sur plusieurs annees',
     [(1, 1, '2017-01-01', 1), (1, 1, '2016-12-31', 2), (1, 1, '2018-06-30', 3)],
     [(1, '2016-12-31')]),
    ('Plusieurs joueurs, lignes melangees',
     [(3, 1, '2019-01-03', 1), (1, 1, '2019-01-02', 1), (2, 1, '2019-01-01', 1), (1, 2, '2019-01-01', 1), (3, 1, '2019-01-01', 1), (2, 2, '2019-01-05', 1)],
     [(1, '2019-01-01'), (2, '2019-01-01'), (3, '2019-01-01')]),
    ('Meme date de premiere connexion pour tous',
     [(1, 1, '2021-01-01', 1), (2, 2, '2021-01-01', 2), (3, 3, '2021-01-01', 3), (3, 3, '2021-01-09', 4)],
     [(1, '2021-01-01'), (2, '2021-01-01'), (3, '2021-01-01')]),
    ('Identifiants de joueurs non consecutifs',
     [(10, 1, '2020-01-02', 1), (500, 1, '2020-01-01', 1), (500, 2, '2020-01-03', 1), (9000, 3, '2019-12-31', 1)],
     [(10, '2020-01-02'), (500, '2020-01-01'), (9000, '2019-12-31')]),
    ('Dates du meme mois',
     [(1, 1, '2020-03-10', 1), (1, 1, '2020-03-02', 1), (1, 1, '2020-03-31', 1)],
     [(1, '2020-03-02')]),
    ('Table vide',
     [],
     []),
    ("Joueur qui change d'appareil a chaque connexion",
     [(5, 1, '2022-01-01', 1), (5, 2, '2022-01-02', 1), (5, 3, '2022-01-03', 1)],
     [(5, '2022-01-01')])
]

# Grands jeux de donnees
debut = date(2016, 1, 1)
# 1) 1000 joueurs x 5 connexions (5000 lignes), inserees de la plus recente a la plus ancienne ;
#    la premiere connexion du joueur p est le 1er janvier 2016 + (p % 30) jours.
gros_1 = [(p, 1 + k, (debut + timedelta(days=p % 30 + 10 * k)).isoformat(), k)
          for p in range(1, 1001) for k in range(4, -1, -1)]
attendu_1 = [(p, (debut + timedelta(days=p % 30)).isoformat()) for p in range(1, 1001)]
cas_tests.append(("Grand jeu : 1000 joueurs, 5 connexions chacun, lignes a l'envers", gros_1, attendu_1))
# 2) un seul joueur avec 1000 connexions, inserees dans l'ordre decroissant des dates
gros_2 = [(7, 1, (debut + timedelta(days=j)).isoformat(), j) for j in range(999, -1, -1)]
cas_tests.append(("Grand jeu : un joueur, 1000 connexions", gros_2, [(7, debut.isoformat())]))

cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def court(liste, n=3):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


for nom, activite, attendu in cas_tests:
    conn = creer_base(activite)
    detail = ""
    try:
        colonnes, lignes = executer(requete, conn)
        if colonnes != COLONNES_ATTENDUES:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendues : {COLONNES_ATTENDUES}"
        else:
            obtenu = Counter((l[0], str(l[1])) for l in lignes)
            voulu = Counter((p, str(d)) for p, d in attendu)
            manquantes = sorted((voulu - obtenu).elements(), key=str)
            en_trop = sorted((obtenu - voulu).elements(), key=str)
            ok = not manquantes and not en_trop
            morceaux = []
            if manquantes:
                morceaux.append(f"Lignes manquantes : {court(manquantes)}")
            if en_trop:
                morceaux.append(f"Lignes en trop (ou date incorrecte) : {court(en_trop)}")
            detail = " | ".join(morceaux)
        nb_obtenu = len(lignes)
    except Exception as erreur:
        ok, nb_obtenu, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}'>{len(activite)}</td>"
        f"<td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Lignes de Activity</th><th style="{th}">Lignes attendues</th>
<th style="{th}">Lignes obtenues</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
