```python
import sqlite3
from IPython.display import HTML, display


def creer_base(employes):
    """Cree une base SQLite en memoire avec la table Employee(id, name, salary, managerId).
    `employes` est une liste de tuples (id, name, salary, managerId)."""
    conn = sqlite3.connect(":memory:")
    conn.execute("CREATE TABLE Employee (id INTEGER PRIMARY KEY, name TEXT, salary INTEGER, managerId INTEGER)")
    conn.executemany("INSERT INTO Employee VALUES (?, ?, ?, ?)", employes)
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange)."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{v}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base([
    (1, "Joe", 70000, 3),
    (2, "Henry", 80000, 4),
    (3, "Sam", 60000, None),
    (4, "Max", 90000, None),
])

afficher(*executer("SELECT * FROM Employee", base_exemple), titre="Table Employee")
```

```python
import html
from collections import Counter

COLONNE_ATTENDUE = "Employee"

# Chaque cas : (nom, lignes de la table Employee (id, name, salary, managerId), noms attendus dans le resultat)
cas_tests = [
    ('Exemple 1',
     [(1, 'Joe', 70000, 3), (2, 'Henry', 80000, 4), (3, 'Sam', 60000, None), (4, 'Max', 90000, None)],
     ['Joe']),
    ("Aucun employe n'a de manager",
     [(1, 'A', 100, None), (2, 'B', 200, None)],
     []),
    ('Salaire egal a celui du manager',
     [(1, 'Boss', 100, None), (2, 'Egal', 100, 1)],
     []),
    ('Employe gagnant moins que son manager',
     [(1, 'Boss', 100, None), (2, 'Moins', 50, 1)],
     []),
    ('Plusieurs employes gagnent plus',
     [(1, 'Boss', 100, None), (2, 'A', 150, 1), (3, 'B', 120, 1), (4, 'C', 90, 1)],
     ['A', 'B']),
    ('Manager inexistant dans la table',
     [(1, 'Orphelin', 500, 99), (2, 'Autre', 10, None)],
     []),
    ('Hierarchie a trois niveaux',
     [(1, 'A', 100, 2), (2, 'B', 80, 3), (3, 'C', 90, None)],
     ['A']),
    ('Noms identiques (pas de dedoublonnage)',
     [(1, 'Boss', 100, None), (2, 'Alex', 150, 1), (3, 'Alex', 130, 1)],
     ['Alex', 'Alex']),
    ('Employe qui est son propre manager',
     [(1, 'Moi', 100, 1), (2, 'Autre', 50, None)],
     []),
    ('Un seul employe',
     [(1, 'Solo', 100, None)],
     []),
    ('Table vide',
     [],
     []),
    ('Identifiants non consecutifs',
     [(10, 'X', 300, 500), (500, 'Y', 200, None), (700, 'Z', 100, 500)],
     ['X']),
    ('Deux managers differents',
     [(1, 'M1', 100, None), (2, 'M2', 300, None), (3, 'E1', 200, 1), (4, 'E2', 200, 2)],
     ['E1']),
    ('Salaires tres grands',
     [(1, 'Boss', 2147483646, None), (2, 'Elu', 2147483647, 1)],
     ['Elu'])
]

# Grands jeux de donnees (1000 employes)
# 1) chaine hierarchique : l'employe i a pour manager l'employe i - 1 et gagne plus que lui
chaine = [(1, "E1", 10, None)] + [(i, f"E{i}", i * 10, i - 1) for i in range(2, 1001)]
cas_tests.append(("Grand jeu : chaine de 1000 employes, salaires croissants", chaine, [f"E{i}" for i in range(2, 1001)]))
# 2) memes liens, salaires decroissants : personne ne gagne plus que son manager
chaine_inverse = [(1, "E1", 10000, None)] + [(i, f"E{i}", 10000 - i * 10, i - 1) for i in range(2, 1001)]
cas_tests.append(("Grand jeu : chaine de 1000 employes, salaires decroissants", chaine_inverse, []))
# 3) un manager (salaire 500) et 999 employes de salaire egal a leur identifiant : seuls ceux > 500 gagnent plus
etoile = [(1, "M", 500, None)] + [(i, f"E{i}", i, 1) for i in range(2, 1001)]
cas_tests.append(("Grand jeu : un manager, 999 employes", etoile, [f"E{i}" for i in range(501, 1001)]))

cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def court(liste, n=5):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


for nom, employes, attendu in cas_tests:
    conn = creer_base(employes)
    detail = ""
    try:
        colonnes, lignes = executer(requete, conn)
        if colonnes != [COLONNE_ATTENDUE]:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendue : ['{COLONNE_ATTENDUE}']"
        else:
            obtenu = Counter(l[0] for l in lignes)
            voulu = Counter(attendu)
            manquants = sorted((voulu - obtenu).elements(), key=str)
            en_trop = sorted((obtenu - voulu).elements(), key=str)
            ok = not manquants and not en_trop
            morceaux = []
            if manquants:
                morceaux.append(f"Noms manquants : {court(manquants)}")
            if en_trop:
                morceaux.append(f"Noms en trop : {court(en_trop)}")
            detail = " | ".join(morceaux)
        nb_obtenu = len(lignes)
    except Exception as erreur:
        ok, nb_obtenu, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}'>{len(employes)}</td>"
        f"<td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Employes</th><th style="{th}">Lignes attendues</th>
<th style="{th}">Lignes obtenues</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
