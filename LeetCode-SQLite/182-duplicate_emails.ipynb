```python
import sqlite3
from IPython.display import HTML, display


def creer_base(emails):
    """Cree une base SQLite en memoire avec la table Person(id, email).
    `emails` est la liste des adresses ; les identifiants sont 1, 2, 3, ..."""
    conn = sqlite3.connect(":memory:")
    conn.execute("CREATE TABLE Person (id INTEGER PRIMARY KEY, email TEXT NOT NULL)")
    conn.executemany("INSERT INTO Person VALUES (?, ?)", list(enumerate(emails, 1)))
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange)."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{v}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base(["a@b.com", "c@d.com", "a@b.com"])

afficher(*executer("SELECT * FROM Person", base_exemple), titre="Table Person")
```

```python
import html
from collections import Counter

COLONNE_ATTENDUE = "Email"

# Chaque cas : (nom, adresses de la table Person dans l'ordre des id, adresses en double attendues (une fois chacune))
cas_tests = [
    ('Exemple 1',
     ['a@b.com', 'c@d.com', 'a@b.com'],
     ['a@b.com']),
    ('Aucun doublon',
     ['a@b.com', 'c@d.com', 'e@f.com'],
     []),
    ('Une adresse presente trois fois',
     ['x@y.com', 'x@y.com', 'z@y.com', 'x@y.com'],
     ['x@y.com']),
    ('Deux adresses en double',
     ['a@b.com', 'c@d.com', 'a@b.com', 'c@d.com', 'e@f.com'],
     ['a@b.com', 'c@d.com']),
    ('Toutes les adresses identiques',
     ['a@b.com', 'a@b.com', 'a@b.com'],
     ['a@b.com']),
    ('Une seule ligne',
     ['solo@mail.com'],
     []),
    ('Table vide',
     [],
     []),
    ('Doublons consecutifs',
     ['d@d.com', 'd@d.com', 'u@u.com'],
     ['d@d.com']),
    ('Doublons eloignes',
     ['d@d.com', 'a@a.com', 'b@b.com', 'c@c.com', 'd@d.com'],
     ['d@d.com']),
    ('Adresses proches mais differentes',
     ['a@b.com', 'a@b.co', 'a@bb.com', 'aa@b.com'],
     []),
    ('Meme partie locale, domaines differents',
     ['jean@a.com', 'jean@b.com', 'jean@a.com'],
     ['jean@a.com']),
    ('Doublon unique parmi de nombreuses adresses',
     ['m1@x.com', 'm2@x.com', 'm3@x.com', 'm4@x.com', 'm2@x.com', 'm5@x.com'],
     ['m2@x.com']),
    ('Adresses avec chiffres et symboles',
     ['a.b+1@mail.com', 'a.b+1@mail.com', 'a.b+2@mail.com', 'a_b@mail.com'],
     ['a.b+1@mail.com'])
]

# Grands jeux de donnees (1000 lignes)
paires = [f"user{i // 2}@mail.com" for i in range(1000)]                   # 500 adresses, chacune presente deux fois
cas_tests += [
    ("Grand jeu : 500 adresses presentes deux fois", paires, [f"user{k}@mail.com" for k in range(500)]),
    ("Grand jeu : 1000 adresses toutes differentes", [f"user{i}@mail.com" for i in range(1000)], []),
    ("Grand jeu : 1000 fois la meme adresse", ["same@mail.com"] * 1000, ["same@mail.com"]),
]

cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def court(liste, n=5):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


for nom, emails, attendu in cas_tests:
    conn = creer_base(emails)
    detail = ""
    try:
        colonnes, lignes = executer(requete, conn)
        if colonnes != [COLONNE_ATTENDUE]:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendue : ['{COLONNE_ATTENDUE}']"
        else:
            obtenu = Counter(l[0] for l in lignes)
            voulu = Counter(attendu)
            manquants = sorted((voulu - obtenu).elements())
            en_trop = sorted((obtenu - voulu).elements())
            ok = not manquants and not en_trop
            morceaux = []
            if manquants:
                morceaux.append(f"Adresses manquantes : {court(manquants)}")
            if en_trop:
                morceaux.append(f"Adresses en trop (ou repetees) : {court(en_trop)}")
            detail = " | ".join(morceaux)
        nb_obtenu = len(lignes)
    except Exception as erreur:
        ok, nb_obtenu, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}'>{len(emails)}</td>"
        f"<td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Lignes dans la table</th><th style="{th}">Lignes attendues</th>
<th style="{th}">Lignes obtenues</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
