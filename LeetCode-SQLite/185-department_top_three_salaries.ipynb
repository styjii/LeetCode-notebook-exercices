<div style="border-left:4px solid #2f6fad;background:rgba(47,111,173,0.08);padding:14px 20px;border-radius:0 8px 8px 0;line-height:1.7;">
Les dirigeants d'une entreprise souhaitent savoir qui gagne le plus dans chacun des d&eacute;partements. Un <b>gros salaire</b> (<i>high earner</i>) d'un d&eacute;partement est un employ&eacute; dont le salaire fait partie des <b>trois salaires <u>uniques</u> les plus &eacute;lev&eacute;s</b> de ce d&eacute;partement.
</div>
<div style="margin-top:14px;line-height:1.7;">&Eacute;crivez une requ&ecirc;te SQL qui trouve les employ&eacute;s qui sont des gros salaires dans chacun des d&eacute;partements. Retournez le tableau r&eacute;sultat dans <b>n'importe quel ordre</b>.</div>
<div style="margin-top:14px;display:flex;flex-wrap:wrap;gap:12px;"><div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Colonnes attendues</div>
<div style="margin-top:4px;line-height:1.6;">Dans cet ordre : <code>Department</code> (nom du d&eacute;partement), <code>Employee</code> (nom de l'employ&eacute;) et <code>Salary</code> (son salaire).</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Salaires uniques</div>
<div style="margin-top:4px;line-height:1.6;">On classe les valeurs de salaire <b>distinctes</b> de chaque d&eacute;partement. Plusieurs employ&eacute;s au m&ecirc;me salaire partagent le m&ecirc;me rang.</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Plus de trois lignes possibles</div>
<div style="margin-top:4px;line-height:1.6;">Un d&eacute;partement peut avoir plus de trois employ&eacute;s dans le r&eacute;sultat si des salaires sont &agrave; &eacute;galit&eacute;.</div></div>

```python
import sqlite3
from IPython.display import HTML, display


def creer_base(departements, employes):
    """Cree une base SQLite en memoire avec les tables Department(id, name) et
    Employee(id, name, salary, departmentId).
    `departements` : liste de tuples (id, name) ; `employes` : liste de tuples (id, name, salary, departmentId)."""
    conn = sqlite3.connect(":memory:")
    conn.executescript(
        "CREATE TABLE Department (id INTEGER PRIMARY KEY, name TEXT NOT NULL);"
        "CREATE TABLE Employee (id INTEGER PRIMARY KEY, name TEXT, salary INTEGER, "
        "departmentId INTEGER REFERENCES Department(id));"
    )
    conn.executemany("INSERT INTO Department VALUES (?, ?)", departements)
    conn.executemany("INSERT INTO Employee VALUES (?, ?, ?, ?)", employes)
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange)."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{v}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base(
    departements=[(1, "IT"), (2, "Sales")],
    employes=[(1, "Joe", 85000, 1), (2, "Henry", 80000, 2), (3, "Sam", 60000, 2), (4, "Max", 90000, 1),
              (5, "Janet", 69000, 1), (6, "Randy", 85000, 1), (7, "Will", 70000, 1)],
)

afficher(*executer("SELECT * FROM Employee", base_exemple), titre="Table Employee")
afficher(*executer("SELECT * FROM Department", base_exemple), titre="Table Department")
```

```python
import html
from collections import Counter

COLONNES_ATTENDUES = ["Department", "Employee", "Salary"]

# Chaque cas : (nom, departements [(id, name)], employes [(id, name, salary, departmentId)],
#               lignes attendues [(Department, Employee, Salary), ...])
cas_tests = [
    ('Exemple 1',
     [(1, 'IT'), (2, 'Sales')],
     [(1, 'Joe', 85000, 1), (2, 'Henry', 80000, 2), (3, 'Sam', 60000, 2), (4, 'Max', 90000, 1), (5, 'Janet', 69000, 1), (6, 'Randy', 85000, 1), (7, 'Will', 70000, 1)],
     [('IT', 'Joe', 85000), ('Sales', 'Henry', 80000), ('Sales', 'Sam', 60000), ('IT', 'Max', 90000), ('IT', 'Randy', 85000), ('IT', 'Will', 70000)]),
    ('Moins de trois employes dans le departement',
     [(1, 'RH')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1)],
     [('RH', 'A', 100), ('RH', 'B', 90)]),
    ('Exactement trois salaires distincts',
     [(1, 'IT')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 80, 1)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 80)]),
    ('Quatre salaires distincts : le plus faible est exclu',
     [(1, 'IT')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 80, 1), (4, 'D', 70, 1)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 80)]),
    ('Egalite en tete (RANK et ROW_NUMBER echouent)',
     [(1, 'IT')],
     [(1, 'A', 100, 1), (2, 'B', 100, 1), (3, 'C', 90, 1), (4, 'D', 80, 1), (5, 'E', 70, 1)],
     [('IT', 'A', 100), ('IT', 'B', 100), ('IT', 'C', 90), ('IT', 'D', 80)]),
    ('Egalite sur le troisieme salaire',
     [(1, 'IT')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 80, 1), (4, 'D', 80, 1), (5, 'E', 80, 1), (6, 'F', 70, 1)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 80), ('IT', 'D', 80), ('IT', 'E', 80)]),
    ('Egalite sur le deuxieme salaire',
     [(1, 'IT')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 90, 1), (4, 'D', 80, 1), (5, 'E', 70, 1)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 90), ('IT', 'D', 80)]),
    ('Un seul salaire distinct, plusieurs employes',
     [(1, 'IT')],
     [(1, 'A', 100, 1), (2, 'B', 100, 1), (3, 'C', 100, 1), (4, 'D', 100, 1)],
     [('IT', 'A', 100), ('IT', 'B', 100), ('IT', 'C', 100), ('IT', 'D', 100)]),
    ('Deux departements independants',
     [(1, 'IT'), (2, 'RH')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 80, 1), (4, 'D', 70, 1), (5, 'E', 10, 2), (6, 'F', 5, 2)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 80), ('RH', 'E', 10), ('RH', 'F', 5)]),
    ('Departement sans employe',
     [(1, 'IT'), (2, 'Vide')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1)],
     [('IT', 'A', 100), ('IT', 'B', 90)]),
    ('Salaires identiques entre departements',
     [(1, 'IT'), (2, 'RH')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 80, 1), (4, 'D', 70, 1), (5, 'E', 70, 2), (6, 'F', 60, 2), (7, 'G', 50, 2), (8, 'H', 40, 2)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 80), ('RH', 'E', 70), ('RH', 'F', 60), ('RH', 'G', 50)]),
    ('Meme nom, salaires differents dans un departement',
     [(1, 'IT')],
     [(1, 'Alex', 100, 1), (2, 'Alex', 90, 1), (3, 'Bob', 80, 1), (4, 'Cal', 70, 1)],
     [('IT', 'Alex', 100), ('IT', 'Alex', 90), ('IT', 'Bob', 80)]),
    ('Entree non triee',
     [(2, 'RH'), (1, 'IT')],
     [(7, 'G', 50, 1), (6, 'F', 95, 2), (5, 'E', 80, 1), (4, 'D', 95, 2), (3, 'C', 90, 1), (2, 'B', 70, 2), (1, 'A', 60, 1)],
     [('RH', 'F', 95), ('IT', 'E', 80), ('RH', 'D', 95), ('IT', 'C', 90), ('RH', 'B', 70), ('IT', 'A', 60)]),
    ('Salaire nul parmi les trois premiers',
     [(1, 'IT')],
     [(1, 'A', 20, 1), (2, 'B', 10, 1), (3, 'C', 0, 1), (4, 'D', 0, 1)],
     [('IT', 'A', 20), ('IT', 'B', 10), ('IT', 'C', 0), ('IT', 'D', 0)]),
    ('Aucun employe',
     [(1, 'IT')],
     [],
     []),
    ('Identifiants non consecutifs',
     [(10, 'IT'), (200, 'RH')],
     [(7, 'A', 5, 10), (70, 'B', 9, 200), (700, 'C', 9, 200), (7000, 'D', 3, 10), (7001, 'E', 4, 10), (7002, 'F', 1, 10)],
     [('IT', 'A', 5), ('RH', 'B', 9), ('RH', 'C', 9), ('IT', 'D', 3), ('IT', 'E', 4)]),
    ("Salaire du 4e rang egal au 3e d'un autre departement",
     [(1, 'IT'), (2, 'RH')],
     [(1, 'A', 100, 1), (2, 'B', 90, 1), (3, 'C', 80, 1), (4, 'D', 50, 1), (5, 'E', 50, 2), (6, 'F', 40, 2), (7, 'G', 30, 2)],
     [('IT', 'A', 100), ('IT', 'B', 90), ('IT', 'C', 80), ('RH', 'E', 50), ('RH', 'F', 40), ('RH', 'G', 30)])
]

# Grands jeux de donnees (1000 employes)
deps_10 = [(d, f"D{d}") for d in range(1, 11)]
# 1) 10 departements ; l'employe i est dans le departement (i % 10) + 1 et gagne i.
#    Les 3 plus grands salaires de chaque departement sont ceux des identifiants 971 a 1000.
emp_1000 = [(i, f"E{i}", i, i % 10 + 1) for i in range(1, 1001)]
cas_tests.append(("Grand jeu : 10 departements, 1000 salaires distincts", deps_10, emp_1000,
                  [(f"D{i % 10 + 1}", f"E{i}", i) for i in range(971, 1001)]))
# 2) un seul departement ; salaires 0 a 4 repetes : les 3 salaires uniques retenus sont 4, 3 et 2
emp_rep = [(i, f"E{i}", i % 5, 1) for i in range(1, 1001)]
cas_tests.append(("Grand jeu : 1000 employes, 5 salaires distincts", [(1, "Unique")], emp_rep,
                  [("Unique", f"E{i}", i % 5) for i in range(1, 1001) if i % 5 >= 2]))

cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def court(liste, n=3):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


for nom, departements, employes, attendu in cas_tests:
    conn = creer_base(departements, employes)
    detail = ""
    try:
        colonnes, lignes = executer(requete, conn)
        if colonnes != COLONNES_ATTENDUES:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendues : {COLONNES_ATTENDUES}"
        else:
            obtenu = Counter(tuple(l) for l in lignes)
            voulu = Counter(attendu)
            manquantes = sorted((voulu - obtenu).elements(), key=str)
            en_trop = sorted((obtenu - voulu).elements(), key=str)
            ok = not manquantes and not en_trop
            morceaux = []
            if manquantes:
                morceaux.append(f"Lignes manquantes : {court(manquantes)}")
            if en_trop:
                morceaux.append(f"Lignes en trop : {court(en_trop)}")
            detail = " | ".join(morceaux)
        nb_obtenu = len(lignes)
    except Exception as erreur:
        ok, nb_obtenu, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}'>{len(departements)} / {len(employes)}</td>"
        f"<td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Departements / Employes</th><th style="{th}">Lignes attendues</th>
<th style="{th}">Lignes obtenues</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
