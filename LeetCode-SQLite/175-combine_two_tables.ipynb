```python
import sqlite3
from IPython.display import HTML, display


def creer_base(personnes, adresses):
    """Cree une base SQLite en memoire contenant les tables Person et Address."""
    conn = sqlite3.connect(":memory:")
    conn.executescript(
        "CREATE TABLE Person (personId INTEGER PRIMARY KEY, lastName TEXT, firstName TEXT);"
        "CREATE TABLE Address (addressId INTEGER PRIMARY KEY, personId INTEGER, city TEXT, state TEXT);"
    )
    conn.executemany("INSERT INTO Person VALUES (?, ?, ?)", personnes)
    conn.executemany("INSERT INTO Address VALUES (?, ?, ?, ?)", adresses)
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange)."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"
    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{v}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base(
    personnes=[(1, "Wang", "Allen"), (2, "Alice", "Bob")],
    adresses=[(1, 2, "New York City", "New York"), (2, 3, "Leetcode", "California")],
)

afficher(*executer("SELECT * FROM Person", base_exemple), titre="Table Person")
afficher(*executer("SELECT * FROM Address", base_exemple), titre="Table Address")
```

```python
import html
from collections import Counter

COLONNES_ATTENDUES = ["firstName", "lastName", "city", "state"]

# Chaque cas : (nom, lignes de Person, lignes de Address, resultat attendu)
# Person : (personId, lastName, firstName)   Address : (addressId, personId, city, state)
cas_tests = [
    ('Exemple 1',
     [(1, 'Wang', 'Allen'), (2, 'Alice', 'Bob')],
     [(1, 2, 'New York City', 'New York'), (2, 3, 'Leetcode', 'California')],
     [('Allen', 'Wang', None, None), ('Bob', 'Alice', 'New York City', 'New York')]),
    ('Toutes les personnes ont une adresse',
     [(1, 'Martin', 'Anna'), (2, 'Durand', 'Luc')],
     [(1, 1, 'Paris', 'IDF'), (2, 2, 'Lyon', 'ARA')],
     [('Anna', 'Martin', 'Paris', 'IDF'), ('Luc', 'Durand', 'Lyon', 'ARA')]),
    ('Aucune adresse',
     [(1, 'Martin', 'Anna'), (2, 'Durand', 'Luc')],
     [],
     [('Anna', 'Martin', None, None), ('Luc', 'Durand', None, None)]),
    ('Une seule personne sans adresse',
     [(7, 'Solo', 'Han')],
     [],
     [('Han', 'Solo', None, None)]),
    ("Adresse d'une personne inexistante uniquement",
     [(1, 'Martin', 'Anna')],
     [(1, 99, 'Ghost', 'Nowhere')],
     [('Anna', 'Martin', None, None)]),
    ('Adresses orphelines melangees',
     [(1, 'A', 'Aa'), (3, 'C', 'Cc')],
     [(1, 3, 'Nice', 'PACA'), (2, 2, 'Orphan', 'X'), (3, 5, 'Orphan2', 'Y')],
     [('Aa', 'A', None, None), ('Cc', 'C', 'Nice', 'PACA')]),
    ('Personne sans adresse au milieu',
     [(1, 'A', 'Aa'), (2, 'B', 'Bb'), (3, 'C', 'Cc')],
     [(1, 1, 'Rome', 'Lazio'), (2, 3, 'Turin', 'Piemonte')],
     [('Aa', 'A', 'Rome', 'Lazio'), ('Bb', 'B', None, None), ('Cc', 'C', 'Turin', 'Piemonte')]),
    ('Meme ville pour plusieurs personnes',
     [(1, 'A', 'Aa'), (2, 'B', 'Bb'), (3, 'C', 'Cc')],
     [(1, 1, 'Oslo', 'Oslo'), (2, 2, 'Oslo', 'Oslo'), (3, 3, 'Oslo', 'Oslo')],
     [('Aa', 'A', 'Oslo', 'Oslo'), ('Bb', 'B', 'Oslo', 'Oslo'), ('Cc', 'C', 'Oslo', 'Oslo')]),
    ('Adresse existante avec ville NULL',
     [(1, 'A', 'Aa'), (2, 'B', 'Bb')],
     [(1, 1, None, 'Texas')],
     [('Aa', 'A', None, 'Texas'), ('Bb', 'B', None, None)]),
    ('Noms identiques, identifiants differents',
     [(1, 'Dupont', 'Jean'), (2, 'Dupont', 'Jean')],
     [(1, 2, 'Lille', 'HDF')],
     [('Jean', 'Dupont', None, None), ('Jean', 'Dupont', 'Lille', 'HDF')]),
    ('Identifiants non consecutifs',
     [(10, 'X', 'Xx'), (200, 'Y', 'Yy'), (3000, 'Z', 'Zz')],
     [(5, 200, 'Brest', 'BRE'), (6, 3000, 'Metz', 'GES')],
     [('Xx', 'X', None, None), ('Yy', 'Y', 'Brest', 'BRE'), ('Zz', 'Z', 'Metz', 'GES')]),
    ("Table Person vide", [], [(1, 1, "Paris", "IDF")], [])
]

# Grand jeu de donnees : 1000 personnes, adresse pour les identifiants pairs uniquement
grand_person = [(i, f"L{i}", f"F{i}") for i in range(1, 1001)]
grand_address = [(i, i, f"C{i}", f"S{i}") for i in range(2, 1001, 2)]
grand_attendu = [(f"F{i}", f"L{i}", f"C{i}", f"S{i}") if i % 2 == 0 else (f"F{i}", f"L{i}", None, None)
                 for i in range(1, 1001)]
cas_tests.append(("Grand jeu : 1000 personnes, moitie sans adresse", grand_person, grand_address, grand_attendu))

cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def court(liste, n=3):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


for nom, personnes, adresses, attendu in cas_tests:
    conn = creer_base(personnes, adresses)
    detail = ""
    try:
        colonnes, lignes = executer(requete, conn)
        lignes = [tuple(l) for l in lignes]
        if colonnes != COLONNES_ATTENDUES:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendues : {COLONNES_ATTENDUES}"
        else:
            obtenu, voulu = Counter(lignes), Counter(attendu)
            manquantes = list((voulu - obtenu).elements())
            en_trop = list((obtenu - voulu).elements())
            ok = not manquantes and not en_trop
            if not ok:
                detail = (f"Lignes manquantes : {court(manquantes)}" if manquantes else "") + \
                         (" | " if manquantes and en_trop else "") + \
                         (f"Lignes en trop : {court(en_trop)}" if en_trop else "")
        nb_obtenu = len(lignes)
    except Exception as erreur:
        ok, nb_obtenu, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}'>{len(personnes)} / {len(adresses)}</td>"
        f"<td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Person / Address</th><th style="{th}">Lignes attendues</th>
<th style="{th}">Lignes obtenues</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
