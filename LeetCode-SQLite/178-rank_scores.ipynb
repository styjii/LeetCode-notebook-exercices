```python
import sqlite3
from IPython.display import HTML, display


def creer_base(scores):
    """Cree une base SQLite en memoire avec la table Scores(id, score).
    `scores` est la liste des scores ; les identifiants sont 1, 2, 3, ..."""
    conn = sqlite3.connect(":memory:")
    conn.execute("CREATE TABLE Scores (id INTEGER PRIMARY KEY, score REAL)")
    conn.executemany("INSERT INTO Scores VALUES (?, ?)", list(enumerate(scores, 1)))
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange, nombres reels a 2 decimales)."""
    cellule = "padding:6px 14px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"

    def texte(v):
        return f"{v:.2f}" if isinstance(v, float) else str(v)

    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{texte(v)}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base([3.50, 3.65, 4.00, 3.85, 4.00, 3.65])

afficher(*executer("SELECT * FROM Scores", base_exemple), titre="Table Scores")
```

```python
import html
from collections import Counter

COLONNES_ATTENDUES = ["score", "rank"]

# Chaque cas : (nom, liste des scores, resultat attendu [(score, rang), ...])
cas_tests = [
    ('Exemple 1', [3.5, 3.65, 4.0, 3.85, 4.0, 3.65], [(4.0, 1), (4.0, 1), (3.85, 2), (3.65, 3), (3.65, 3), (3.5, 4)]),
    ('Un seul score', [2.5], [(2.5, 1)]),
    ('Table vide', [], []),
    ('Tous les scores identiques', [3.0, 3.0, 3.0], [(3.0, 1), (3.0, 1), (3.0, 1)]),
    ('Tous les scores distincts', [1.0, 3.0, 2.0], [(3.0, 1), (2.0, 2), (1.0, 3)]),
    ('Egalite en bas du classement', [5.0, 4.0, 4.0], [(5.0, 1), (4.0, 2), (4.0, 2)]),
    ('Deux egalites consecutives (pas de trou)', [10.0, 9.0, 10.0, 8.0, 9.0], [(10.0, 1), (10.0, 1), (9.0, 2), (9.0, 2), (8.0, 3)]),
    ('Egalite au milieu', [9.0, 8.0, 8.0, 7.0], [(9.0, 1), (8.0, 2), (8.0, 2), (7.0, 3)]),
    ('Scores proches : zero et 0.01', [0.0, 0.01], [(0.01, 1), (0.0, 2)]),
    ('Entree non triee avec triples egalites', [3.1, 3.4, 3.1, 3.9, 3.4, 3.9, 3.1], [(3.9, 1), (3.9, 1), (3.4, 2), (3.4, 2), (3.1, 3), (3.1, 3), (3.1, 3)]),
    ('Decimales proches', [1.1, 1.01, 1.1, 1.01], [(1.1, 1), (1.1, 1), (1.01, 2), (1.01, 2)]),
    ('Scores superieurs a 10', [99.99, 100.0, 99.99, 0.5], [(100.0, 1), (99.99, 2), (99.99, 2), (0.5, 3)])
]

# Grands jeux de donnees (1000 lignes)
# 1) 50 scores distincts (0.00 a 0.49), chacun repete 20 fois : le rang du score k/100 est 50 - k
valeurs = [(i % 50) / 100 for i in range(1000)]
attendu_50 = [((49 - k) / 100, k + 1) for k in range(50) for _ in range(20)]
cas_tests.append(("Grand jeu : 50 scores distincts repetes 20 fois", valeurs, attendu_50))
# 2) 1000 scores tous distincts, melanges : le score i/100 a le rang 1001 - i
melange = [((i * 7919) % 1000 + 1) / 100 for i in range(1000)]
attendu_1000 = [(i / 100, 1001 - i) for i in range(1000, 0, -1)]
cas_tests.append(("Grand jeu : 1000 scores distincts melanges", melange, attendu_1000))


def arrondi(lignes):
    """Arrondit les scores a 2 decimales pour comparer des nombres reels."""
    return [(round(float(s), 2), r) for s, r in lignes]


def court(liste, n=3):
    texte = ", ".join(html.escape(repr(x)) for x in liste[:n])
    return texte + (f", ... (+{len(liste) - n})" if len(liste) > n else "")


cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0

for nom, scores, attendu in cas_tests:
    conn = creer_base(scores)
    detail = ""
    try:
        colonnes, lignes = executer(requete, conn)
        if colonnes != COLONNES_ATTENDUES:
            ok = False
            detail = f"Colonnes obtenues : {html.escape(str(colonnes))} ; attendues : {COLONNES_ATTENDUES}"
        else:
            obtenu = arrondi(lignes)
            voulu = arrondi(attendu)
            manquantes = list((Counter(voulu) - Counter(obtenu)).elements())
            en_trop = list((Counter(obtenu) - Counter(voulu)).elements())
            tri_ok = all(a[0] >= b[0] for a, b in zip(obtenu, obtenu[1:]))
            ok = not manquantes and not en_trop and tri_ok
            morceaux = []
            if manquantes:
                morceaux.append(f"Lignes manquantes : {court(manquantes)}")
            if en_trop:
                morceaux.append(f"Lignes en trop : {court(en_trop)}")
            if not manquantes and not en_trop and not tri_ok:
                morceaux.append("Valeurs correctes, mais le resultat n'est pas trie par score decroissant")
            detail = " | ".join(morceaux)
        nb_obtenu = len(lignes)
    except Exception as erreur:
        ok, nb_obtenu, detail = False, "-", f"Erreur : {html.escape(str(erreur))}"
    reussis += ok
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    apercu = str([f"{s:.2f}" for s in scores]).replace("'", "") if len(scores) <= 8 else f"{len(scores)} scores"
    lignes_html.append(
        f"<tr><td style='{cell}'>{nom}</td>"
        f"<td style='{cell}font-family:monospace;'>{apercu}</td>"
        f"<td style='{cell}'>{len(attendu)}</td><td style='{cell}'>{nb_obtenu}</td>"
        f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;"
        f"font-size:12px;font-weight:600;'>{etat}</span></td>"
        f"<td style='{cell}font-size:12px;font-family:monospace;'>{detail}</td></tr>")

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Scores</th><th style="{th}">Lignes attendues</th>
<th style="{th}">Lignes obtenues</th><th style="{th}">Statut</th><th style="{th}">Detail</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
