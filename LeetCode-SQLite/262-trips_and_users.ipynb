<div style="border-left:4px solid #2f6fad;background:rgba(47,111,173,0.08);padding:14px 20px;border-radius:0 8px 8px 0;line-height:1.7;">
Le <b>taux d'annulation</b> se calcule en divisant le nombre de demandes <b>annul&eacute;es</b> (par le client ou par le chauffeur) avec des utilisateurs non bannis, par le <b>nombre total</b> de demandes avec des utilisateurs non bannis ce jour-l&agrave;.
</div>
<div style="margin-top:14px;line-height:1.7;">&Eacute;crivez une requ&ecirc;te SQL qui trouve le <b>taux d'annulation</b> des demandes avec des utilisateurs non bannis (<b>le client et le chauffeur ne doivent pas &ecirc;tre bannis</b>) pour chaque jour entre <code>"2013-10-01"</code> et <code>"2013-10-03"</code> comptant <b>au moins un trajet</b>. Arrondissez <code>Cancellation Rate</code> &agrave; <b>deux d&eacute;cimales</b>. Retournez le tableau r&eacute;sultat dans <b>n'importe quel ordre</b>.</div>
<div style="margin-top:14px;display:flex;flex-wrap:wrap;gap:12px;"><div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Colonnes attendues</div>
<div style="margin-top:4px;line-height:1.6;">Dans cet ordre : <code>Day</code> (la date) et <code>Cancellation Rate</code> (avec une espace dans le nom).</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Trajets retenus</div>
<div style="margin-top:4px;line-height:1.6;">Seuls comptent les trajets dont le client <b>et</b> le chauffeur ne sont pas bannis, dont la date est comprise entre le 2013-10-01 et le 2013-10-03 inclus.</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Taux</div>
<div style="margin-top:4px;line-height:1.6;">Annul&eacute;s (par le client ou par le chauffeur) divis&eacute;s par le total des trajets retenus du jour, arrondi &agrave; deux d&eacute;cimales.</div></div>

```python
import sqlite3
from IPython.display import HTML, display


def creer_base(utilisateurs, trajets):
    """Cree une base SQLite en memoire avec les tables Users et Trips.
    `utilisateurs` : liste de tuples (users_id, banned, role)
    `trajets` : liste de tuples (id, client_id, driver_id, city_id, status, request_at)"""
    conn = sqlite3.connect(":memory:")
    conn.executescript(
        "CREATE TABLE Users (users_id INTEGER PRIMARY KEY, "
        "banned TEXT CHECK (banned IN ('Yes', 'No')), "
        "role TEXT CHECK (role IN ('client', 'driver', 'partner')));"
        "CREATE TABLE Trips (id INTEGER PRIMARY KEY, client_id INTEGER, driver_id INTEGER, city_id INTEGER, "
        "status TEXT CHECK (status IN ('completed', 'cancelled_by_driver', 'cancelled_by_client')), "
        "request_at TEXT);"
    )
    conn.executemany("INSERT INTO Users VALUES (?, ?, ?)", utilisateurs)
    conn.executemany("INSERT INTO Trips VALUES (?, ?, ?, ?, ?, ?)", trajets)
    conn.commit()
    conn.execute("PRAGMA query_only = ON")        # la requete ne peut pas modifier les donnees
    return conn


def executer(requete, conn):
    """Execute une requete et retourne (noms_des_colonnes, lignes)."""
    curseur = conn.execute(requete)
    if curseur.description is None:
        raise ValueError("la requete ne retourne aucune colonne (requete vide ?)")
    colonnes = [d[0] for d in curseur.description]
    return colonnes, curseur.fetchall()


def afficher(colonnes, lignes, titre=""):
    """Affiche un resultat sous forme de tableau (NULL en orange, nombres reels a 2 decimales)."""
    cellule = "padding:6px 12px;border-bottom:1px solid rgba(128,128,128,0.35);font-family:Consolas,monospace;"

    def texte(v):
        return f"{v:.2f}" if isinstance(v, float) else str(v)

    corps = "".join(
        "<tr>" + "".join(
            f"<td style='{cellule}color:#d9892b;font-style:italic;'>NULL</td>" if v is None
            else f"<td style='{cellule}'>{texte(v)}</td>" for v in ligne) + "</tr>"
        for ligne in lignes)
    entete = "".join(f"<th style='{cellule}text-align:left;'>{c}</th>" for c in colonnes)
    display(HTML(f"<div style='font-family:Segoe UI,Helvetica,Arial,sans-serif;'><b>{titre}</b>"
                 f"<table style='border-collapse:collapse;border:1px solid rgba(128,128,128,0.35);margin:4px 0;'>"
                 f"<tr style='background:rgba(47,111,173,0.15);'>{entete}</tr>{corps}</table>"
                 f"<div style='opacity:0.7;font-size:12px;'>{len(lignes)} ligne(s)</div></div>"))


# Donnees de l'exemple 1
base_exemple = creer_base(
    utilisateurs=[(1, "No", "client"), (2, "Yes", "client"), (3, "No", "client"), (4, "No", "client"),
                  (10, "No", "driver"), (11, "No", "driver"), (12, "No", "driver"), (13, "No", "driver")],
    trajets=[(1, 1, 10, 1, "completed", "2013-10-01"), (2, 2, 11, 1, "cancelled_by_driver", "2013-10-01"),
             (3, 3, 12, 6, "completed", "2013-10-01"), (4, 4, 13, 6, "cancelled_by_client", "2013-10-01"),
             (5, 1, 10, 1, "completed", "2013-10-02"), (6, 2, 11, 6, "completed", "2013-10-02"),
             (7, 3, 12, 6, "completed", "2013-10-02"), (8, 2, 12, 12, "completed", "2013-10-03"),
             (9, 3, 10, 12, "completed", "2013-10-03"), (10, 4, 13, 12, "cancelled_by_driver", "2013-10-03")],
)

afficher(*executer("SELECT * FROM Trips", base_exemple), titre="Table Trips")
afficher(*executer("SELECT * FROM Users", base_exemple), titre="Table Users")
```
