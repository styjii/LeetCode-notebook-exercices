```python
import ast
import inspect
import textwrap
from IPython.display import HTML, display

# Chaque cas : (nom, x, resultat attendu)
cas_tests = [
    ('Exemple 1', 121, True),
    ('Exemple 2', -121, False),
    ('Exemple 3', 10, False),
    ('Zero', 0, True),
    ('Un chiffre', 7, True),
    ('Deux chiffres identiques', 11, True),
    ('Deux chiffres differents', 12, False),
    ('Negatif a un chiffre', -1, False),
    ('Negatif palindromique en apparence', -101, False),
    ('Longueur paire, palindrome', 1221, True),
    ('Longueur impaire, palindrome', 12321, True),
    ('Longueur paire, non palindrome', 1231, False),
    ('Longueur impaire, non palindrome', 123421, False),
    ('Se termine par zero', 100, False),
    ('Zeros au milieu, palindrome', 1001, True),
    ('Zeros au milieu, non palindrome', 1000021, False),
    ('Multiple de 10', 20, False),
    ('Debut et fin identiques, milieu different', 1234567891, False),
    ('Palindrome de 10 chiffres (grand)', 1410110141, True),
    ('Palindrome proche de la borne haute', 2147447412, True),
    ('Borne haute', 2147483647, False),
    ('Borne basse', -2147483648, False),
    ('Un suivi de zeros', 1000000000, False),
    ('Presque palindrome (dernier chiffre different)', 1000000002, False),
    ('Palindrome 1000000001', 1000000001, True),
    ('Palindrome 1230321', 1230321, True),
    ('Palindrome 9999', 9999, True),
    ('Presque palindrome 12344321 modifie', 12354321, False),
    ('Palindrome 12344321', 12344321, True),
]

sol = Solution()
cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes, reussis = [], 0


def ligne(nom, x, attendu, obtenu, ok):
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    return (f"<tr><td style='{cell}'>{nom}</td>"
            f"<td style='{cell}font-family:monospace;'>{x}</td>"
            f"<td style='{cell}font-family:monospace;'>{attendu}</td>"
            f"<td style='{cell}font-family:monospace;'>{obtenu}</td>"
            f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;"
            f"border-radius:12px;font-size:12px;font-weight:600;'>{etat}</span></td></tr>")


for nom, x, attendu in cas_tests:
    try:
        obtenu = sol.isPalindrome(x)
        ok = type(obtenu) is bool and obtenu == attendu          # le type doit etre bool
    except Exception as erreur:
        obtenu, ok = f"Erreur : {erreur}", False
    reussis += ok
    lignes.append(ligne(nom, x, attendu, obtenu, ok))

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"


# ---- Indication sur la question de suivi : conversion en chaine de caracteres ? ----
def conversions_chaine(source):
    """Liste les conversions en chaine detectees dans le code source (analyse syntaxique)."""
    trouves = []
    for noeud in ast.walk(ast.parse(textwrap.dedent(source))):
        if isinstance(noeud, ast.Call):
            f = noeud.func
            nom_f = f.id if isinstance(f, ast.Name) else (f.attr if isinstance(f, ast.Attribute) else None)
            if nom_f in {"str", "repr", "format"}:
                trouves.append(f"{nom_f}() ligne {noeud.lineno}")
        elif isinstance(noeud, ast.JoinedStr):
            trouves.append(f"f-string ligne {noeud.lineno}")
    return trouves


try:
    trouves = conversions_chaine(inspect.getsource(Solution.isPalindrome))
    if trouves:
        msg_suivi, couleur_suivi = ("Conversion en chaine detectee (" + ", ".join(trouves) + "). "
                                    "Solution valide, mais la question de suivi demande de s'en passer."), "#d9892b"
    else:
        msg_suivi, couleur_suivi = "Aucune conversion en chaine detectee : question de suivi respectee.", "#2e9e5b"
except Exception:
    msg_suivi, couleur_suivi = "Verification automatique de la question de suivi impossible dans cet environnement.", "#7f8c8d"

th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">x</th><th style="{th}">Attendu</th>
<th style="{th}">Obtenu</th><th style="{th}">Statut</th></tr>
{''.join(lignes)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div>
<div style="margin-top:10px;padding:10px 16px;border-left:4px solid {couleur_suivi};background:rgba(128,128,128,0.08);line-height:1.6;">
<b>Question de suivi :</b> {msg_suivi}
</div></div>"""))
```

<div style="opacity:0.8;margin-bottom:10px;">Cliquez sur une piste pour l'afficher. Elles sont progressives.</div>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 1</summary>
<div style="margin-top:8px;line-height:1.6;">Certains nombres peuvent &ecirc;tre rejet&eacute;s imm&eacute;diatement, sans aucun calcul. Que dire des nombres n&eacute;gatifs ? Et des nombres qui se terminent par 0 (sauf 0 lui-m&ecirc;me) ?</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 2</summary>
<div style="margin-top:8px;line-height:1.6;">Pour la version sans cha&icirc;ne : comment extraire le dernier chiffre d'un entier, et comment le retirer de l'entier ? Pensez au reste et au quotient d'une division par 10.</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">

<div style="margin-top:8px;line-height:1.6;">Vous pouvez inverser tout le nombre puis le comparer &agrave; l'original. Mais en 32 bits, l'inversion peut d&eacute;passer l'intervalle autoris&eacute; : existe-t-il une fa&ccedil;on d'&eacute;viter ce risque ?</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">

Comment savoir que vous avez atteint la moiti&eacute; sans conna&icirc;tre le nombre de chiffres ? Observez l'&eacute;volution de la partie restante et de la partie invers&eacute;e.</div></details>

<div style="margin-top:8px;line-height:1.6;">Une fois la moiti&eacute; atteinte, comparez les deux parties. Que faut-il faire diff&eacute;remment lorsque le nombre a un nombre impair de chiffres (le chiffre du milieu) ?</div></details>
