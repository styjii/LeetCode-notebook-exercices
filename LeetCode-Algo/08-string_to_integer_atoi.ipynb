<div style="font-size:12px;letter-spacing:2px;text-transform:uppercase;opacity:0.75;">Exercice d'algorithmique &nbsp;|&nbsp; LeetCode n&deg;8</div>
<div style="font-size:32px;font-weight:700;margin:8px 0 14px 0;">String to Integer (atoi)</div>
<div style="font-size:15px;opacity:0.9;">Conversion d'une cha&icirc;ne de caract&egrave;res en entier sign&eacute; sur 32 bits</div>
<div style="margin-top:18px;">
<span style="background:#d9892b;color:#fff;padding:4px 14px;border-radius:14px;font-size:12px;font-weight:600;margin-right:8px;">Moyen</span>
<span style="background:rgba(255,255,255,0.18);color:#fff;padding:4px 14px;border-radius:14px;font-size:12px;">Cha&icirc;nes de caract&egrave;res</span>
</div></div>


<div style="margin:28px 0 12px 0;padding-bottom:8px;border-bottom:2px solid #1f4e79;">
<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">1. &Eacute;nonc&eacute;</span>
</div>
<div style="border-left:4px solid #2f6fad;background:rgba(47,111,173,0.08);padding:14px 20px;border-radius:0 8px 8px 0;line-height:1.7;">
Impl&eacute;mentez la fonction <code>myAtoi(string s)</code>, qui convertit une cha&icirc;ne de caract&egrave;res en un <b>entier sign&eacute; sur 32 bits</b>.
</div>
<div style="margin:14px 0 8px 0;font-weight:700;">L'algorithme de <code>myAtoi</code> est le suivant :</div>
<div style="display:flex;align-items:flex-start;margin-bottom:10px;">
<span style="flex:0 0 28px;height:28px;line-height:28px;text-align:center;border-radius:50%;background:#1f4e79;color:#fff;font-weight:700;font-size:13px;margin-right:14px;">1</span>
<span style="line-height:1.6;padding-top:2px;"><b>Espaces :</b> ignorez tous les espaces en d&eacute;but de cha&icirc;ne (<code>' '</code>).</span></div>
<div style="display:flex;align-items:flex-start;margin-bottom:10px;">
<span style="flex:0 0 28px;height:28px;line-height:28px;text-align:center;border-radius:50%;background:#1f4e79;color:#fff;font-weight:700;font-size:13px;margin-right:14px;">2</span>
<span style="line-height:1.6;padding-top:2px;"><b>Signe :</b> v&eacute;rifiez si le caract&egrave;re suivant est <code>'-'</code> ou <code>'+'</code> et d&eacute;terminez le signe du r&eacute;sultat ; le r&eacute;sultat est positif si aucun des deux n'est pr&eacute;sent.</span></div>
<div style="display:flex;align-items:flex-start;margin-bottom:10px;">
<span style="flex:0 0 28px;height:28px;line-height:28px;text-align:center;border-radius:50%;background:#1f4e79;color:#fff;font-weight:700;font-size:13px;margin-right:14px;">3</span>
<span style="line-height:1.6;padding-top:2px;"><b>Conversion :</b> lisez les chiffres en ignorant les z&eacute;ros en t&ecirc;te, jusqu'&agrave; rencontrer un caract&egrave;re qui n'est pas un chiffre ou la fin de la cha&icirc;ne. Si aucun chiffre n'a &eacute;t&eacute; lu, le r&eacute;sultat est <code>0</code>.</span></div>
<div style="display:flex;align-items:flex-start;margin-bottom:10px;">
<span style="flex:0 0 28px;height:28px;line-height:28px;text-align:center;border-radius:50%;background:#1f4e79;color:#fff;font-weight:700;font-size:13px;margin-right:14px;">4</span>
<span style="line-height:1.6;padding-top:2px;"><b>Arrondi :</b> si l'entier sort de l'intervalle <b>[&minus;2<sup>31</sup>, 2<sup>31</sup> &minus; 1]</b>, ramenez-le dans l'intervalle : les valeurs inf&eacute;rieures &agrave; &minus;2<sup>31</sup> deviennent &minus;2<sup>31</sup>, et celles sup&eacute;rieures &agrave; 2<sup>31</sup> &minus; 1 deviennent 2<sup>31</sup> &minus; 1.</span></div>

<div style="margin-top:6px;line-height:1.7;">Retournez l'entier obtenu comme r&eacute;sultat final.</div>
<div style="margin-top:14px;border:1px solid #c0392b;background:rgba(192,57,43,0.08);border-radius:8px;padding:12px 18px;line-height:1.7;">
<span style="font-weight:700;color:#c0392b;text-transform:uppercase;letter-spacing:1px;font-size:12px;">Interdiction</span><br>
Vous ne devez utiliser <b>aucune fonction int&eacute;gr&eacute;e</b> qui convertit une cha&icirc;ne en nombre (par exemple <code>atoi</code>, <code>stoi</code>, <code>parseInt</code>, <code>int()</code>, <code>Number()</code>). Effectuez la conversion <b>manuellement</b>.
</div>
