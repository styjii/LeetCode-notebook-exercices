<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">5. Votre solution</span>
</div>


```python
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        for i in range(0, len(nums)):
            curr_val = nums[i]
            for j in range(i+1, len(nums)):
                next_val = nums[j]
                
                if curr_val + next_val == target:
                    return [i, j]
```

<div style="margin:28px 0 12px 0;padding-bottom:8px;border-bottom:2px solid #1f4e79;">
<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">6. Tests</span>
</div>


```python
import random
import time
from IPython.display import HTML, display


def verifier(nums, target, resultat):
    """Verifie qu'un resultat est valide (l'ordre des indices n'a pas d'importance)."""
    if not isinstance(resultat, (list, tuple)) or len(resultat) != 2:
        return False
    i, j = resultat
    return (0 <= i < len(nums) and 0 <= j < len(nums)
            and i != j and nums[i] + nums[j] == target)


cas_tests = [
    ("Exemple 1", [2, 7, 11, 15], 9),
    ("Exemple 2", [3, 2, 4], 6),
    ("Exemple 3", [3, 3], 6),
    ("Nombres negatifs", [-1, -2, -3, -4, -5], -8),
    ("Zeros", [0, 4, 3, 0], 0),
    ("Valeurs extremes", [-10**9, 10**9, 5], 0),
]

sol = Solution()
lignes, reussis = [], 0


def ligne(nom, entree, resultat, ok, duree=""):
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    return (f"<tr><td style='padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);'>{nom}</td>"
            f"<td style='padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);font-family:monospace;'>{entree}</td>"
            f"<td style='padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);font-family:monospace;'>{resultat}</td>"
            f"<td style='padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);'>{duree}</td>"
            f"<td style='padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);'>"
            f"<span style='background:{couleur};color:#fff;padding:2px 12px;border-radius:12px;font-size:12px;font-weight:600;'>{etat}</span></td></tr>")


for nom, nums, target in cas_tests:
    res = sol.twoSum(list(nums), target)
    ok = verifier(nums, target, res)
    reussis += ok
    lignes.append(ligne(nom, f"nums={nums}, target={target}", res, ok))

# Test de performance : n = 10 000
n = 10_000
nums = random.sample(range(-10**9, 10**9), n)
i, j = random.sample(range(n), 2)
target = nums[i] + nums[j]
debut = time.perf_counter()
res = sol.twoSum(list(nums), target)
duree = time.perf_counter() - debut
ok = verifier(nums, target, res)
reussis += ok
lignes.append(ligne("Performance", "n = 10 000", res, ok, f"{duree*1000:.1f} ms"))

total = len(cas_tests) + 1
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="padding:8px 12px;">Test</th><th style="padding:8px 12px;">Entree</th>
<th style="padding:8px 12px;">Resultat</th><th style="padding:8px 12px;">Duree</th><th style="padding:8px 12px;">Statut</th></tr>
{''.join(lignes)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
</div></div>"""))
```
