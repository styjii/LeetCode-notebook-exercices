```python
from typing import Optional


class ListNode:
    """Noeud d'une liste chainee simple (identique a la definition de LeetCode)."""

    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

<div style="margin:28px 0 12px 0;padding-bottom:8px;border-bottom:2px solid #1f4e79;">
<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">6. Fonctions utilitaires</span>
</div>
<div style="opacity:0.85;line-height:1.6;">Ces fonctions servent uniquement &agrave; construire et afficher des listes pour vos essais personnels. Elles sont utilis&eacute;es par la cellule de tests.</div>


```python
def construire(valeurs):
    """Construit une liste chainee a partir d'une liste Python : [2, 4, 3] -> 2 -> 4 -> 3."""
    tete = ListNode()
    courant = tete
    for v in valeurs:
        courant.next = ListNode(v)
        courant = courant.next
    return tete.next


def en_liste(noeud, limite=1000):
    """Convertit une liste chainee en liste Python (la limite protege contre les cycles)."""
    valeurs = []
    while noeud is not None and len(valeurs) < limite:
        valeurs.append(noeud.val)
        noeud = noeud.next
    return valeurs


def afficher(noeud):
    """Affiche une liste chainee sous la forme 2 -> 4 -> 3 -> None."""
    print(" -> ".join(map(str, en_liste(noeud))) + " -> None")


# Exemple d'utilisation
afficher(construire([2, 4, 3]))
```

<div style="margin:28px 0 12px 0;padding-bottom:8px;border-bottom:2px solid #1f4e79;">
<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">7. Votre solution</span>
</div>


```python
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        result = ListNode()
        content, rest = result, 0
        
        while l1 or l2 or rest:
            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            total = val1 + val2 + rest
            rest = total // 10

            content.next = ListNode(total % 10)
            content = content.next

            if l1: l1 = l1.next
            if l2: l2 = l2.next
        return result.next
```
