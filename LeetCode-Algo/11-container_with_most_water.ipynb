<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">1. &Eacute;nonc&eacute;</span>
</div>
<div style="border-left:4px solid #2f6fad;background:rgba(47,111,173,0.08);padding:14px 20px;border-radius:0 8px 8px 0;line-height:1.7;">
On vous donne un tableau d'entiers <code>height</code> de longueur <code>n</code>. Il y a <code>n</code> lignes verticales dessin&eacute;es de sorte que les deux extr&eacute;mit&eacute;s de la <i>i</i>-&egrave;me ligne soient <code>(i, 0)</code> et <code>(i, height[i])</code>.<br>
Trouvez deux lignes qui, avec l'axe des abscisses, forment un <b>conteneur</b> pouvant contenir le <b>plus d'eau possible</b>.
</div>
<div style="margin-top:14px;line-height:1.7;">Retournez la <b>quantit&eacute; maximale d'eau</b> qu'un conteneur peut stocker.</div>
<div style="margin-top:14px;border:1px solid #c98a1b;background:rgba(201,138,27,0.10);border-radius:8px;padding:12px 18px;line-height:1.7;">
<span style="font-weight:700;color:#c98a1b;text-transform:uppercase;letter-spacing:1px;font-size:12px;">Remarque de l'&eacute;nonc&eacute;</span><br>
Vous <b>ne pouvez pas incliner</b> le conteneur.
</div>
<div style="margin-top:14px;display:flex;flex-wrap:wrap;gap:12px;"><div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Aire d'un conteneur</div>
<div style="margin-top:4px;line-height:1.6;">Pour les lignes d'indices <code>i &lt; j</code> : <code>min(height[i], height[j]) &times; (j &minus; i)</code>. La largeur est l'&eacute;cart des indices, la hauteur d'eau est limit&eacute;e par la ligne <b>la plus courte</b>.</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Deux lignes</div>
<div style="margin-top:4px;line-height:1.6;">Un conteneur est d&eacute;fini par exactement deux lignes. Les lignes situ&eacute;es entre elles ne comptent pas (elles sont immerg&eacute;es).</div></div>
<div style="flex:1;min-width:220px;border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:12px 16px;">
<div style="font-size:11px;text-transform:uppercase;letter-spacing:1px;opacity:0.65;">Sortie</div>
<div style="margin-top:4px;line-height:1.6;">Un entier : l'aire maximale. Elle vaut 0 si toutes les lignes ont une hauteur nulle.</div></div>
</div>

```python
import time
from IPython.display import HTML, display

# Chaque cas : (nom, hauteurs, aire maximale attendue). Les valeurs attendues sont ecrites explicitement.
cas_tests = [
    ('Exemple 1', [1, 8, 6, 2, 5, 4, 8, 3, 7], 49),
    ('Exemple 2', [1, 1], 1),
    ('Deux lignes de hauteurs differentes', [2, 1], 1),
    ('Deux lignes nulles', [0, 0], 0),
    ('Une ligne nulle entre deux lignes egales', [5, 0, 5], 10),
    ('Trois lignes', [1, 2, 1], 2),
    ('Lignes aux extremites les plus hautes', [4, 3, 2, 1, 4], 16),
    ('Hauteurs croissantes', [1, 2, 4, 3], 4),
    ('Hauteurs decroissantes', [5, 4, 3, 2, 1], 6),
    ('Meilleur conteneur au milieu', [2, 3, 4, 5, 18, 17, 6], 17),
    ('Extremites hautes, centre bas', [1000, 1, 1, 1, 1000], 4000),
    ('Hauteur maximale des deux lignes', [10000, 10000], 10000),
    ('Une seule ligne non nulle', [0, 0, 7, 0, 0], 0),
    ('Plateau de hauteurs identiques', [3, 3, 3, 3, 3, 3], 15),
    ('Pic isole', [1, 1, 1, 100, 1, 1, 1], 6),
    ('Alternance haut / bas', [9, 1, 9, 1, 9, 1, 9], 54)
]


def suite_pseudo_aleatoire(n, graine):
    """Suite deterministe de n hauteurs entre 0 et 10 000 (generateur congruentiel lineaire)."""
    x, resultat = graine, []
    for _ in range(n):
        x = (x * 1103515245 + 12345) % (2 ** 31)
        resultat.append((x >> 8) % 10001)
    return resultat


# Grands jeux de donnees dont la reponse est connue
N = 100_000
escalier = list(range(10001))                      # 0, 1, 2, ..., 10000 : meilleur conteneur = 5000 x 5000
pics = [0] * N
pics[1000], pics[99000] = 10000, 10000             # deux pics de hauteur 10 000, distants de 98 000
cas_tests += [
    ("Aleatoire (n = 2 000)", suite_pseudo_aleatoire(2000, 1), 19_055_193),
    ("Escalier croissant (n = 10 001)", escalier, 25_000_000),
    ("Escalier decroissant (n = 10 001)", escalier[::-1], 25_000_000),
    ("Toutes les hauteurs a 10 000 (n = 100 000)", [10000] * N, 999_990_000),
    ("Extremites hautes, milieu bas (n = 100 000)", [10000] + [1] * (N - 2) + [10000], 999_990_000),
    ("Deux pics eloignes (n = 100 000)", pics, 980_000_000),
    ("Toutes les hauteurs nulles (n = 100 000)", [0] * N, 0),
    ("Aleatoire (n = 100 000)", suite_pseudo_aleatoire(N, 2), 992_692_829),
]

sol = Solution()
cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes_html, reussis = [], 0


def apercu(h):
    return str(h) if len(h) <= 10 else f"[{h[0]}, {h[1]}, {h[2]}, ... ] ({len(h)} lignes)"


def ligne(nom, entree, attendu, obtenu, ok, duree):
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    if ok and duree > 1.0:
        couleur, etat = "#d9892b", "LENT"
    return (f"<tr><td style='{cell}'>{nom}</td>"
            f"<td style='{cell}font-family:monospace;'>{entree}</td>"
            f"<td style='{cell}font-family:monospace;'>{attendu:,}</td>"
            f"<td style='{cell}font-family:monospace;'>{obtenu}</td>"
            f"<td style='{cell}'>{duree*1000:.1f} ms</td>"
            f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;"
            f"border-radius:12px;font-size:12px;font-weight:600;'>{etat}</span></td></tr>")


for nom, hauteurs, attendu in cas_tests:
    copie = list(hauteurs)
    debut = time.perf_counter()
    try:
        obtenu = sol.maxArea(copie)
        ok = type(obtenu) is int and obtenu == attendu
    except Exception as erreur:
        obtenu, ok = f"Erreur : {erreur}", False
    duree = time.perf_counter() - debut
    reussis += ok
    lignes_html.append(ligne(nom, apercu(hauteurs), attendu, obtenu, ok, duree))

total = len(cas_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Entree</th><th style="{th}">Attendu</th>
<th style="{th}">Obtenu</th><th style="{th}">Duree</th><th style="{th}">Statut</th></tr>
{''.join(lignes_html)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
<span style="opacity:0.7;"> &nbsp;(le statut LENT signale une reponse correcte obtenue en plus d'une seconde)</span>
</div></div>"""))
```

<span style="font-size:20px;font-weight:700;letter-spacing:0.3px;color:#2f6fad;">8. Pistes de r&eacute;flexion</span>
</div>
<div style="opacity:0.8;margin-bottom:10px;">Cliquez sur une piste pour l'afficher. Elles sont progressives.</div>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 1</summary>
<div style="margin-top:8px;line-height:1.6;">Commencez par la formule : pour deux lignes d'indices <code>i &lt; j</code>, l'aire vaut <code>min(height[i], height[j]) &times; (j &minus; i)</code>. Tester toutes les paires donne la bonne r&eacute;ponse, mais en O(n<sup>2</sup>) : trop lent pour 10<sup>5</sup> lignes.</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 2</summary>
<div style="margin-top:8px;line-height:1.6;">Deux facteurs jouent sur l'aire : la <b>largeur</b> (l'&eacute;cart entre les indices) et la <b>hauteur</b> (la plus courte des deux lignes). Quel est le conteneur de largeur maximale ? Que se passe-t-il si vous le prenez comme point de d&eacute;part ?</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 3</summary>
<div style="margin-top:8px;line-height:1.6;">Placez un pointeur sur chaque extr&eacute;mit&eacute; et faites-les se rapprocher. &Agrave; chaque &eacute;tape, vous devez d&eacute;cider quel pointeur d&eacute;placer. Quand vous d&eacute;placez un pointeur, la largeur diminue : pour esp&eacute;rer une aire plus grande, sur quel facteur pouvez-vous encore jouer ?</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 4</summary>
<div style="margin-top:8px;line-height:1.6;">L'aire est limit&eacute;e par la ligne la plus courte. Si vous d&eacute;placez le pointeur de la ligne <b>la plus haute</b>, la hauteur ne peut pas augmenter (elle reste born&eacute;e par l'autre ligne) alors que la largeur diminue. Quel pointeur vaut-il donc mieux d&eacute;placer ?</div></details>
<details style="border:1px solid rgba(128,128,128,0.35);border-radius:8px;padding:10px 16px;margin-bottom:8px;">
<summary style="cursor:pointer;font-weight:600;">Piste 5</summary>
<div style="margin-top:8px;line-height:1.6;">Pensez &agrave; m&eacute;moriser la meilleure aire rencontr&eacute;e, &agrave; arr&ecirc;ter la boucle quand les pointeurs se rejoignent, et &agrave; v&eacute;rifier les cas limites : n = 2, hauteurs nulles, hauteurs toutes &eacute;gales, ligne la plus haute au milieu.</div></details>
