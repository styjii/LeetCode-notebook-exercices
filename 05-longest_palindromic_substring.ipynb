```python
import time
from IPython.display import HTML, display

# Pour chaque cas : (nom, chaine, longueur maximale attendue).
# Plusieurs reponses pouvant etre valides, on verifie la validite et la longueur.
cas_tests = [
    ("Exemple 1", "babad", 3),
    ("Exemple 2", "cbbd", 2),
    ("Un seul caractere", "a", 1),
    ("Deux caracteres distincts", "ac", 1),
    ("Deux caracteres identiques", "aa", 2),
    ("Tous identiques", "aaaa", 4),
    ("Chaine entierement palindromique", "abcba", 5),
    ("Palindrome pair au milieu", "forgeeksskeegfor", 10),
    ("Palindrome isole", "abacdfgdcaba", 3),
    ("Aucun palindrome de longueur > 1", "abcd", 1),
    ("Sensibilite a la casse", "Aa", 1),
    ("Chiffres et lettres", "a1b22b1a", 8),
    ("Palindrome en fin de chaine", "xyzabba", 4),
    ("Palindrome en debut de chaine", "abbaxyz", 4),
]

# Grandes entrees (1000 caracteres) dont la reponse est connue par construction
grands_tests = [
    ("Grande entree : 1000 fois 'a'", "a" * 1000, 1000),
    ("Grande entree : alternance 'ab'", "ab" * 500, 999),
    ("Grande entree : 'b' au centre", "a" * 499 + "b" + "a" * 500, 999),
    ("Grande entree : aucun palindrome > 1 car.", "".join(chr(97 + (i * 7) % 26) + chr(65 + (i * 3) % 26) for i in range(500)), 1),
]

sol = Solution()
cell = "padding:7px 12px;border-bottom:1px solid rgba(128,128,128,0.3);"
lignes, reussis = [], 0


def est_palindrome(t):
    return t == t[::-1]


def verifier(s, resultat, longueur_attendue):
    return (isinstance(resultat, str) and len(resultat) == longueur_attendue
            and resultat in s and est_palindrome(resultat))


def apercu(t):
    return repr(t) if len(t) <= 22 else repr(t[:18]) + f" ... ({len(t)} car.)"


def ligne(nom, entree, attendu, obtenu, ok, duree):
    couleur, etat = ("#2e9e5b", "REUSSI") if ok else ("#c0392b", "ECHEC")
    if ok and duree > 1.0:
        couleur, etat = "#d9892b", "LENT"
    return (f"<tr><td style='{cell}'>{nom}</td>"
            f"<td style='{cell}font-family:monospace;'>{entree}</td>"
            f"<td style='{cell}font-family:monospace;'>{attendu}</td>"
            f"<td style='{cell}font-family:monospace;'>{obtenu}</td>"
            f"<td style='{cell}'>{duree*1000:.1f} ms</td>"
            f"<td style='{cell}'><span style='background:{couleur};color:#fff;padding:2px 12px;"
            f"border-radius:12px;font-size:12px;font-weight:600;'>{etat}</span></td></tr>")


# Le dernier grand cas ne contient aucun palindrome de longueur 2 ou 3 : la longueur maximale est donc 1.
chaine_l = grands_tests[-1][1]
assert all(not est_palindrome(chaine_l[i:i + 2]) for i in range(len(chaine_l) - 1))
assert all(not est_palindrome(chaine_l[i:i + 3]) for i in range(len(chaine_l) - 2))

for nom, s, longueur in cas_tests + grands_tests:
    debut = time.perf_counter()
    try:
        obtenu = sol.longestPalindrome(s)
    except Exception as erreur:
        obtenu = f"Erreur : {erreur}"
    duree = time.perf_counter() - debut
    ok = verifier(s, obtenu, longueur)
    attendu_txt = f"longueur {longueur}"
    reussis += ok
    lignes.append(ligne(nom, apercu(s), attendu_txt, apercu(obtenu) if isinstance(obtenu, str) else obtenu, ok, duree))

total = len(cas_tests) + len(grands_tests)
couleur_bilan = "#2e9e5b" if reussis == total else "#c0392b"
th = "padding:8px 12px;"
display(HTML(f"""
<div style="font-family:'Segoe UI',Helvetica,Arial,sans-serif;">
<table style="border-collapse:collapse;width:100%;">
<tr style="background:rgba(47,111,173,0.15);text-align:left;">
<th style="{th}">Test</th><th style="{th}">Entree</th><th style="{th}">Attendu</th>
<th style="{th}">Obtenu</th><th style="{th}">Duree</th><th style="{th}">Statut</th></tr>
{''.join(lignes)}
</table>
<div style="margin-top:14px;padding:10px 16px;border-left:4px solid {couleur_bilan};background:rgba(128,128,128,0.08);">
<b>Bilan :</b> {reussis} / {total} tests reussis
<span style="opacity:0.7;"> &nbsp;(le statut LENT signale une reponse correcte obtenue en plus d'une seconde)</span>
</div></div>"""))
```
